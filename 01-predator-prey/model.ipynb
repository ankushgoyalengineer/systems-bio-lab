```python
#Prey-predator relationship
#Objective - To assess the relationship between prey and predator population in a given ecosystem.
```

```python
#Prey' population can be modeled using the following equation:
#dx/dt = alpha * x - beta * x * y where x is prey population, y is predator population, alpha is the growth rate of prey, and beta is the predation rate.
#Predator population can be modeled using the following equation:
#dy/dt = delta * x * y - c * y where delta is the growth rate of predators per prey eaten, and c is the natural death rate of predators.
```

```python
#Question 1: What happens to the prey and predator population when there are no predators at all?
#Assuming that there are no predators, we can put y = 0. Prey population will grow according to the growth rate until the resources are exhausted.
#dx = alpha * x * dt
#We are assuming alpha = 2, x = 1 as starting population, dt as 0.1
import matplotlib.pyplot as plt
x = 1
x_coordinates = []
y_coordinates = []
for i in range(100):
    alpha = 2
    dt = 0.1
    dx = alpha * x * dt
    x += dx
    y_coordinates.append(i*dt)
    x_coordinates.append(x)
#Plot this on a graph to visualize the growth of prey population over time.


plt.plot(x_coordinates,y_coordinates)
plt.xlabel("Time")
plt.ylabel("Prey Population")
plt.title("Growth of Prey Population (No Predators)")
plt.show()

#Here we are assuming that resources are infinite
```

```python
#However, even in no predator situation, the prey population will stagnate after some time since resources are finite
#To introduce the resource constraint, we can modify the equation to logistic groqwth equation:
#dx/dt = alpha * x * (1 - x/K) where K is the carrying capacity of the environment

import matplotlib.pyplot as plt
x = 1
x_coordinates = []
y_coordinates = []
for i in range(100):
    alpha = 2
    dt = 0.1
    dx = alpha * x * dt * (1- x/1000)
    x += dx
    y_coordinates.append(i*dt)
    x_coordinates.append(x)
#Plot this on a graph to visualize the growth of prey population over time.


plt.plot(x_coordinates,y_coordinates)
plt.xlabel("Time")
plt.ylabel("Prey Population")
plt.title("Growth of Prey Population (No Predators)")
plt.show()
```

```python
#Question 2: What happens to the predator population when there's no prey to eat — obviously it declines, but how fast, and does it ever fully die out or just get very small?
#dy/dt = delta * x * y - c * y where delta is the growth rate of predators per prey eaten, and c is the natural death rate of predators.

import matplotlib.pyplot as plt
y = 1000
x_coordinates = []
y_coordinates = []
for i in range(100):
    c = 2
    dt = 0.1
    dy = - c * y * dt
    y += dy
    y_coordinates.append(i*dt)
    x_coordinates.append(y)
    if y <= 1:
        print(f"Predator population has died out at time {i*dt}")
        break
#Plot this on a graph to visualize the growth of prey population over time.


plt.plot(x_coordinates,y_coordinates)
plt.xlabel("Time")
plt.ylabel("Predator Population")
plt.title("Growth of Predator Population (No Prey)")
plt.show()
```

```python
#Question 3: What happens to the prey and predator population when there are both prey and predators in the ecosystem?
#Equation we will use here - dy/dx = (delta * x * y - c * y) / (alpha * x - beta * x * y) - It's called phase plane analysis

import matplotlib.pyplot as plt
x = 10
y = 5
x_coordinates = []
y_coordinates = []
time_coordinates = []
current_time = 0
dt = 0.001
c = 1
delta = 0.1
alpha = 1
beta = 0.1

for i in range(10000):

    x_coordinates.append(x)
    y_coordinates.append(y)
    time_coordinates.append(current_time)

    dy = (delta * x * y * dt) - (c * y * dt)
    dx = (alpha * x * dt) - (beta * x * y * dt)

    y += dy
    x += dx
    current_time += dt
    if y <= 0:
        print(f"Predator population has died out at iteration {i}")
        break
    elif x <= 0:
        print(f"Prey population is completely wiped out at iteration {i}")
        break
#Plot this on a graph to visualize the growth of prey population over time.


plt.plot(time_coordinates, x_coordinates, label="Prey Population", color="blue")
plt.plot(time_coordinates, y_coordinates, label="Predator Population", color="red")

plt.xlabel("Time")
plt.ylabel("Population Size")
plt.title("Predator vs Prey Populations Over Time")
plt.legend() # This adds a key to the graph using the labels above
plt.show()
```

```python
#Question 4: What happens to the prey and predator population when there are both prey and predators in the ecosystem as well as limited resources.


import matplotlib.pyplot as plt
x = 10
y = 5
x_coordinates = []
y_coordinates = []
time_coordinates = []
current_time = 0
dt = 0.001
c = 1
delta = 0.1
alpha = 1
beta = 0.1

K = 25

for i in range(50000):

    x_coordinates.append(x)
    y_coordinates.append(y)
    time_coordinates.append(current_time)

    dy = (delta * x * y * dt) - (c * y * dt)
    dx = (alpha * x * (1 - (x / K)) * dt) - (beta * x * y * dt)

    y += dy
    x += dx
    current_time += dt
    if y <= 0:
        print(f"Predator population has died out at iteration {i}")
        break
    elif x <= 0:
        print(f"Prey population is completely wiped out at iteration {i}")
        break
#Plot this on a graph to visualize the growth of prey population over time.


plt.plot(time_coordinates, x_coordinates, label="Prey Population", color="blue")
plt.plot(time_coordinates, y_coordinates, label="Predator Population", color="red")

plt.xlabel("Time")
plt.ylabel("Population Size")
plt.title("Predator vs Prey Populations Over Time")
plt.legend() # This adds a key to the graph using the labels above
plt.show()
```

```python
import matplotlib.pyplot as plt

def Lotka_Volterra(x, y, dt, c, delta, alpha, beta, K):
    #User provides initial prey population (x) and predator population (y)
    #User provides arguements. Standards are dt = 0.001, c = 1.0, delta = 0.1, alpha = 1.0, beta = 0.1, K = 25.0
    
    if y == 0:
        x_coordinates = []
        y_coordinates = []

        for i in range(10000):
            dx = alpha * x * dt * (1- x/K)
            x += dx
            y_coordinates.append(x)
            x_coordinates.append(i*dt)
        plt.plot(x_coordinates,y_coordinates)
        plt.xlabel("Time")
        plt.ylabel("Prey Population")
        plt.title("Growth of Prey Population (No Predators)")
        plt.show()

    elif x == 0:
        x_coordinates = []
        y_coordinates = []
        for i in range(10000):
            dy = - c * y * dt
            y += dy
            y_coordinates.append(y)
            x_coordinates.append(i*dt)
            if y <= 1:
                print(f"Predator population has died out at time {i*dt}")
                break
        #Plot this on a graph to visualize the growth of prey population over time.


        plt.plot(x_coordinates,y_coordinates)
        plt.xlabel("Time")
        plt.ylabel("Predator Population")
        plt.title("Growth of Predator Population (No Prey)")
        plt.show()
    else:
        x_coordinates = []
        y_coordinates = []
        time_coordinates = []
        current_time = 0.0

        for i in range(10000):

            x_coordinates.append(x)
            y_coordinates.append(y)
            time_coordinates.append(current_time)

            dy = (delta * x * y * dt) - (c * y * dt)
            dx = (alpha * x * (1 - (x / K)) * dt) - (beta * x * y * dt)

            y += dy
            x += dx
            current_time += dt
            if y <= 0:
                print(f"Predator population has died out at iteration {i}")
                break
            elif x <= 0:
                print(f"Prey population is completely wiped out at iteration {i}")
                break
        #Plot this on a graph to visualize the growth of prey population over time.


        plt.plot(time_coordinates, x_coordinates, label="Prey Population", color="blue")
        plt.plot(time_coordinates, y_coordinates, label="Predator Population", color="red")

        plt.xlabel("Time")
        plt.ylabel("Population Size")
        plt.title("Predator vs Prey Populations Over Time")
        plt.legend() # This adds a key to the graph using the labels above
        plt.show()
```

```python
import matplotlib.pyplot as plt

def simulate_lotka_volterra(
    prey_initial: float, 
    predator_initial: float, 
    dt: float = 0.001, 
    c: float = 1.0, 
    delta: float = 0.1, 
    alpha: float = 1.0, 
    beta: float = 0.1, 
    K: float = 25.0,
    steps: int = 10000
):
    """
    Simulates predator-prey dynamics using the Lotka-Volterra equations with logistic growth.
    
    Args:
        prey_initial: Starting prey population.
        predator_initial: Starting predator population.
        dt: Time step for numerical integration.
        c: Predator natural death rate.
        delta: Predator growth efficiency from eating prey.
        alpha: Prey natural growth rate.
        beta: Predation rate (prey hunted per predator).
        K: Environmental carrying capacity for prey.
        steps: Number of iterations for the simulation.
    """
    
    # Initialize tracking lists
    time_data = []
    prey_data = []
    predator_data = []
    
    current_prey = prey_initial
    current_predator = predator_initial
    current_time = 0.0

    # Single simulation loop handling all scenarios natively via the math
    for i in range(steps):
        time_data.append(current_time)
        prey_data.append(current_prey)
        predator_data.append(current_predator)

        # Calculate rates of change
        dy = (delta * current_prey * current_predator * dt) - (c * current_predator * dt)
        dx = (alpha * current_prey * (1 - (current_prey / K)) * dt) - (beta * current_prey * current_predator * dt)

        # Update populations
        current_predator += dy
        current_prey += dx
        current_time += dt

        # Prevent physically impossible negative populations
        if current_predator <= 0:
            current_predator = 0
        if current_prey <= 0:
            current_prey = 0


    return time_data, prey_data, predator_data, predator_initial, prey_initial

def plot_result(time_data, prey_data, predator_data, predator_initial, prey_initial):
    # Determine dynamic title based on starting conditions
    if predator_initial == 0:
        graph_title = "Growth of Prey Population (No Predators)"
    elif prey_initial == 0:
        graph_title = "Growth of Predator Population (No Prey)"
    else:
        graph_title = "Predator vs Prey Populations Over Time"    
    # Plotting the results
    plt.plot(time_data, prey_data, label="Prey Population", color="blue")
    plt.plot(time_data, predator_data, label="Predator Population", color="red")

    plt.xlabel("Time")
    plt.ylabel("Population Size")
    plt.title(graph_title)
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.6) # Added a light grid for visual polish
    plt.show()

def plot_phase_portrait(prey_data, predator_data, title_suffix=""):
    """Plots the predator population directly against the prey population."""
    plt.plot(prey_data, predator_data, color="purple")
    
    plt.xlabel("Prey Population")
    plt.ylabel("Predator Population")
    plt.title(f"Phase Portrait {title_suffix}")
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.show()

def plot_comparison(runs, labels):
    """
    Overlays multiple simulation runs on a single plot.
    'runs' is a list of tuples: [(time_data, prey_data, predator_data), ...]
    """
    # Create a slightly wider figure so the overlaid waves aren't cramped
    plt.figure(figsize=(10, 6)) 
    
    # Define a color scheme so we can tell the runs apart
    run_colors = ['blue', 'orange', 'green', 'purple']
    
    # Loop through each simulation run you provided
    for i in range(len(runs)):
        time = runs[i][0]
        prey = runs[i][1]
        pred = runs[i][2]
        color = run_colors[i % len(run_colors)]
        
        # Plot Prey (Solid line) and Predator (Dashed line) for this run
        plt.plot(time, prey, label=f"Prey ({labels[i]})", linestyle='-', color=color)
        plt.plot(time, pred, label=f"Predator ({labels[i]})", linestyle='--', color=color)

    plt.xlabel("Time")
    plt.ylabel("Population Size")
    plt.title("Ecosystem Dynamics Comparison")
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.show()

```

```python
time_data, prey_data, predator_data, predator_initial, prey_initial = simulate_lotka_volterra(prey_initial= 10, predator_initial= 10, dt= 0.001, c= 1.0, delta = 0.1,     
    alpha = 1.0, 
    beta = 0.1, 
    K = 25.0,
    steps = 10000
)

plot_result(time_data, prey_data, predator_data, predator_initial, prey_initial)
```
